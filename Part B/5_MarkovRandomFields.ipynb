# Markov Random Fields

To solve two-dimensional spatial problems, we extend the temporal axis to a spatial **regular grid** or **lattice**.
To work on MRFs we need to define some core concepts:
- For a regular grid of NxM nodes we have NxM possible Random Variables
- To define the probability of a state in the grid following the Markov property we need to define the concept of **neighborhood**:
$P(x_m|x_{-m}) = P(x_m|x_1, x_2, \dots, x_{m-1}, x_{m+1}, \dots, x_{NM}) = P(x_m|x_{\Delta m})$, where $\Delta m$ is the neighborhood of m.

### Rules for neighbors
- The point is not a neighbor of itself.
- If point A is neighbor of B, then B is the neighbor of A.
- We also need to define the neighborhood structure, the most common are:
    - **4-Neighborhood (Cross)**: up, down, left, right.
    - **8-Neighborhood (Grid)**: includes diagonal points. 

## Clique
- A clique is a subset of the lattice $\Omega, c \in \Omega$ that are fully connected: every node in $c$is connected to every other node in $c$ by an edge.
- The nodes of the clique must be linked to one another.
    - $u$ and $v$ are linked if: 
        - $u = v$
        - $u \in \Delta v$ 
        - $v \in \Delta u$
- The order of the clique is the number of nodes it includes.
    - **Order 1 Clique**: A single point.
    - **Order 2 Clique**: A line connecting two adjecent points.
    - **Order 3 Clique**: Don't exist in the standard 4-neighbor grid, but do exist in the 8-neighbor grid, when including diagonals and forming triangles.
    - etc., depending on the neighborhood structure chosen


### Clique Potential
A cluque potential $U_c(x_c)$ is a local real-valued scoring function assigned to a specific clique $c$.
The input to $U_c(x_c)$ is the subset of variables belonging to clique $c$ (denoted $x_c$). The output is a scalar real number representing the **energy cost** for that specific local combination.

Because probability is proportional to $\exp(-U)$, the conventions are:
- Low potential $U_c$ (low energy): A favorable, plausible, or expected local state $\to$ High probability.
- High potential $U_c$ (high energy): An unfavorable, conflicting, or unnatural local state $\to$ Low probability.

#### Order 1: Single-Site Potentials $U_1(x_i)$
Operates on a single node $i$. It measures how well the chosen state $x_i$ agrees with individual priors or external observations (often called the **data term** or likelihood).
- **Example (Image Denoising)**: Let $y_i$ be the observed noisy pixel brightness, and $x_i$ be the true underlying brighness we want to estimate: $$U_1 = (x_i - y_i)^2$$
    - If the proposed label is close to the noisy sensor reading, $U_1$ is small (low penalty).


#### Order 2: Single-Site Potentials $U_2(x_i, x_j)$
Operates on two directly connected neighbors $\{i, j\}$. It enforces pairwise constraints such as smoothness, correlation, or continuity.

## Gibbs energy

The Gibbs energy is the sum of all the clique potentials.
$$\mathcal{E}(\underline{\mathbf{x}}) = \sum_c U_c(\underline{\mathbf{x}}) 

### Gibbs distribution
$p(\underline{\mathbf{x}}) = A e ^{-\lambda \mathcal{E}(\underline{\mathbf{x}})}$, where $A$ is the normalization constant to ensure that the sum of probabilities of all possible states equals 1. $A = \frac{1}{\int e^{-\lambda \mathcal{E}(\underline{\mathbf{x}})}dx}$

## Hammersley-Clifford Theorem
If the probability distribution of a random field is a Gibbs distribution, then it must be a Markov Random Field, and vice versa.
$$MRF \Leftrightarrow Gibbs$$

Which tells us: we only need to define the potential rules $U_c(x)$ for each clique microscopically, and then add them up, which will yield a field that satisfies the Markov property.

## Practical applications of MRF:
1. **Model a MRF**: Turn your problem into a MRF grid and define potential rules.
2. **Optimization**: Find the state with the highest conditional probability.
    - Solved by using simulated annealing, which requires defining a temperature cooling law, combined with Gibbs sampling technique expoiting the Markov property.
