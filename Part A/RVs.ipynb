## BASICS INTRO DEFINITIONS

### Random Experiment
An experiment that is characterized by the fact that they can result in a set of values, but we don't know before doing the experiment which value in the set will result.

### Random Models
he mathematical formulation used to quantify the experiment. Formally, it is the Kolmogorov triple $(\Omega, \mathcal{F}, P)$:
    - Sample Space ($\Omega$): The set of all possible primitive outcomes $\omega$. (e.g., all infinite possible continuous radiance curves, or $\Omega = \{1, 2, 3, 4, 5, 6\}$).
    - Event Space ($\mathcal{F}$): A $\sigma$-algebra containing valid collections of outcomes (events) to which probabilities can be assigned.
    - Probability Measure ($P$): A function $P: \mathcal{F} \to [0, 1]$ that assigns a probability mass or weight to each event, satisfying $P(\Omega) = 1$ and countable additivity.

###  Random Variables
It is a deterministic function that maps outcomes from the abstract sample space $\Omega$ to real numbers:$$X: \Omega \longrightarrow \mathbb{R} \quad (\text{or } \mathbb{R}^B \text{ for } B\text{-dimensional vectors})$$ Raw outcomes $\omega$ can be anything (e.g., "cloudy sky", "water pixel", "heads/tails"). Mathematics operates on numbers, vectors, and matrices. A random variable translates abstract experimental outcomes into real values so you can do calculus, linear algebra, and sampling.

## Random Variables - cdf and pdf

- A *cdf* or *cumulative distribution function*, denoted as $F_X(x)$, applies both to discrete and continuous RVs. It measures the accumulated probability that the random variable $X$ takes on a value less than or equal to a threshold x:

    $$F_X(x) = P(X \le x) = P(\{\omega \in \Omega \mid X(\omega) \le x\})$$

- a *pdf* or *probability density function* (and *pmf* or *probability mass function* for a discrete RV), is denoted as $f_X(x)$ (or $p_X(x)$) and describes the probability at specific values. For the continuous case the *pdf* is the derivative of the *cdf*.
    - **Discrete case**: for discrete outcomes (e.g., a pixel belonging to class $k \in \{1,...,K\}$): $p_X(x)=P(X=x)$ - $p_X(x)$ represents an **actual probability** and $\sum_{x} p_X(x) = 1$;
    - **Continuous case**: for continuous variables (e.g., reflectance values across spectral wavelengths): $f_X(x) = \frac{d}{dx}F_X(x)$. For a continuous RV, the probability of hitting any exact single number is zero: $P(X=x_0) = 0$ for any single $x_0$, therefore $f_X(x)$ is not a probability - it is a probability density (probability per unit of measurement). $\int_{-\inf}^{+\inf} f_X(x)dx = 1$. To obtain an actual probability, you must inter=grate over an interval $[a,b]: P(a \le X \le b) = F_X(b) - F_X(a) = \int_a^b f_X(x)dx$.

For two random variables, X and Y:
- **Joint distribution $p(X, Y)$**: The probability that X and Y take on specific values **at the same time**. 
- **Marginal distribution $p(X)$ or $p(Y)$**: The distribution of one variable on its own, ignoring the other variable entirely.
- **Conditional Distribution: $p(X \mid Y)$ or $p(Y \mid X)$**: The distribution of one variable after observing the exact state of the other. $p(x \mid y) = \frac{p(x,y)}{p(y)}$ provided $p(y) > 0$.

**Joint = Conditional x Marginal**

$p(x,y) = p(x \mid y)p(y) = p(y \mid x)p(x)$ --> rearranging this identity yields the **Bayes' Rule**: $$p(x \mid y) = \frac{p(y \mid x) \, p(x)}{p(y)}$$


**In MCMC**:
- The numerator $p(y \mid x) p(x)$ is the unnormalized joint distribution $p(x, y)$.
- The denominator $p(y)$ is the intractable marginal distribution (evidence).
- The goal is sampling from the conditional distribution $p(x \mid y)$ (posterior) without needing to compute that marginal integral.
