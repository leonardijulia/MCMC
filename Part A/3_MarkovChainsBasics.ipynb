## Markov Chains Introduction

### State Space & Stochastic Process
- **Stochastic Process**: A collection of random variables $\{X_t\}_{t \in T}$ indexed by time $t \in \{0, 1, 2, \dots\}$ representing the state of a system at each step.
- **State Space ($\mathcal{S}$)**: The set of all possible values (states) that $X_t$ can take. It can be discrete (finite or countably infinite, e.g., $\mathcal{S} = \{s_1, s_2, \dots, s_K\}$) or continuous ($\mathcal{S} \subseteq \mathbb{R}^d$).

### The Markov Property
A sequence of random variables $\{X_t\}$ is a Markov chain if the conditional probability of the future state depends only on the current state, completely independent of the past history: $$P(X_{t+1} = s_j \mid X_t = s_i, X_{t-1} = s_{t-1}, \dots, X_0 = s_0) = P(X_{t+1} = s_j \mid X_t = s_i)$$
- **Time-Homogeneity**: A Markov chain is homogeneous (stationary in time) if the transition probabilities do not change with time step $t$:$$P(X_{t+1} = s_j \mid X_t = s_i) = P_{ij} \quad \text{for all } t$$

A Markov Chain is a random process $\{X(t,w)\}$, where:
- t is discrete $t_n$, $n=0,1,... \rightarrow  X(t_n,w) = X_n(w)$ 
- for each n, X_n is a discrete and finite rv, whose possible values are called states i=1,...,m
- the marginal distribution of each $X_n$ is given by a vector $$\hat{p^{(n)}} = [p_1^{(n)}...p_m^{(n)}]^{T};  p_i^{(n)} \ge 0 \forall i,n; \sum_{i=1}^{m} p_i^{(n)} = 1$$

### Transition Probability Matrix & Kernel
- **Transition Matrix ($P$)**: For a discrete state space of size $K$, the transition probabilities are organized into a $K \times K$ stochastic matrix: $$P = \begin{bmatrix} P_{11} & P_{12} & \dots & P_{1K} \\ P_{21} & P_{22} & \dots & P_{2K} \\ \vdots & \vdots & \ddots & \vdots \\ P_{K1} & P_{K2} & \dots & P_{KK} \end{bmatrix}, \quad P_{ij} = P(X_{t+1} = s_j \mid X_t = s_i)$$ where $s_i$ and $s_j$ represent specific states in the system's state space $S = \{s_1, s_1,..., s_K\}$. The matrix entry $P_{ij}$ defined above can be interpreted as: "Given that the system is currently in state $s_i$, what is the probability that it jumps to state $s_j$ in the next step?"  
    - **Row-Stochastic Condition**: Every row sums to $1$ because moving somewhere in $\mathcal{S}$ is guaranteed: $$\sum_{j \in \mathcal{S}} P_{ij} = 1 \quad \text{and} \quad P_{ij} \ge 0$$
- **Transition Kernel ($T(x, dy)$ or $K(x, x')$)**: For continuous state spaces, the discrete matrix is replaced by a probability density kernel defining the transition density from point $x$ to $x'$: $$\int_\mathcal{S} K(x, x') \, dx' = 1$$

### Classification of the states

To classify states, we first define how they connect:
- **Reachability** ($i \to j$): State $s_j$ is reachable from state $s_i$ (written $i \to j$) if there is a non-zero probability of reaching $s_j$ from $s_i$ in some finite number of steps:$$P_{ij}^{(n)} > 0 \quad \text{for some } n \ge 0$$
- **Communication** ($i \leftrightarrow j$): Two states communicate if $i \to j$ and $j \to i$.

### Essential State
A state $s_i$ is essential if, once the chain reaches $s_i$, it can never wander into a state from which it cannot return to $s_i$.
- ***Formal Definition:** State $s_i$ is essential if for every state $s_j$ such that $i \to j$, it is also true that $j \to i$.
- **Intuition:** There is no point of no return. If you can travel from $s_i$ to anywhere else in the system, there is always a path leading back to $s_i$.

### Inessential State
A state $s_i$ is inessential if there is at least one "one-way door" leading away from it.
- **Formal Definition:** State $s_i$ is inessential if there exists some state $s_k$ such that $i \to k$, but $k \not\to i$ (the chain can reach $s_k$, but cannot return to $s_i$ from $s_k$).
- **Intuition:** The chain can leave $s_i$ and get trapped elsewhere in the system, making a return impossible.

### Periodic MC
Periodicity describes whether a chain is forced into a deterministic, clock-like cycle when returning to a state.

#### Period of a State
For any state $s_i$, consider the set of all time-step lengths $n \ge 1$ for which it is possible to start at $s_i$ and return to $s_i$ with non-zero probability ($P_{ii}^{(n)} > 0$).

The period $d(i)$ of state $s_i$ is the greatest common divisor ($\gcd$) of all those path lengths: $$d(i) = \gcd \{n \ge 1 : P_{ii}^{(n)} > 0\}$$

- **Periodic State:** If $d(i) > 1$, the state is periodic with period $d(i)$. Returns to state $s_i$ can only happen at multiples of $d(i)$ steps (e.g., at steps $2, 4, 6, \dots$ or $3, 6, 9, \dots$).
- **Aperiodic State:** If $d(i) = 1$, the state is aperiodic. The chain can return to state $s_i$ at irregular, non-cyclic intervals.

In an irreducible chain (where all states communicate), all states share the exact same period.
- If that common period is $d > 1$, the entire chain is periodic.
- If that common period is $d = 1$, the entire chain is aperiodic.

## HIA Chains (Homogeneous, Irreducible, Aperiodic)

1. **Homogeneous** - The transition probability matrix $P$ is static across time ($P_{ij}$ does not vary with step $t$).
2. **Irreducible** - Every state can eventually reach every other state ($i \leftrightarrow j$ for all $i, j$). The state space consists of a single communicating class without isolated sinks or inaccessible pockets.
3. **Aperiodic** - The chain has no deterministic cyclic periods ($d = 1$). It cannot get trapped oscillating in fixed rhythm between subsets of states.

The combination of these three properties guarantees the **Fundamental Convergence Theorem for Markov Chains** (in finite state spaces):
   1. *Existence and Uniqueness of the Stationary (Limit) Distribution:* There exists a strictly positive, unique probability distribution vector $\pi = [\pi_1, \dots, \pi_K]$ such that:$$\pi P = \pi \quad \text{and} \quad \sum_{i} \pi_i = 1$$
   2. *Asymptotic Convergence (Loss of Initial Conditions):* Regardless of the initial probability distribution $\pi_0$ at step 0, as the number of steps $n$ approaches infinity:$$\lim_{n \to \infty} \pi_0 P^n = \pi$$ In matrix terms, every row of $P^n$ converges to the exact same stationary vector $\pi$:$$\lim_{n \to \infty} P^n = \begin{bmatrix} — & \pi & — \\ — & \pi & — \\ & \vdots & \\ — & \pi & — \end{bmatrix}$$
   3. *Ergodicity (Time Averages Equal Space Averages):* The fraction of time the chain spends in any state $s_i$ over a long trajectory converges almost surely to $\pi_i$: $$\lim_{N \to \infty} \frac{1}{N} \sum_{t=1}^N \mathbb{I}(X_t = s_i) = \pi_i$$

### *Intuition behind the limit distribution*

In a building with three rooms: Room 1 (Library), Room 2 (Café), and Room 3 (Gym).
1. The Transition Matrix $P$ (The Rules of Motion): Every hour, people move between rooms according to set probabilities:
    - From Library: 70% stay in Library, 20% go to Café, 10% go to Gym.
    - From Café: 30% go to Library, 50% stay in Café, 20% go to Gym.
    - From Gym: 10% go to Library, 40% go to Café, 50% stay in Gym.
2. The Experiment (Initial State $\pi_0$):
At 8:00 AM, 1,000 people enter the building. Suppose every single person starts in Room 1 (Library):$$\pi_0 = [1.0, \, 0.0, \, 0.0]$$ 
3. Watching the Crowd Settle Over Time:
    - Hour 1: People disperse according to row 1 of $P$.
    - Hour 2: People reshuffle again ($\pi_2 = \pi_0 P^2$).
    - Hour 10: People keep moving between rooms individually every hour, but if you step back and count the total heads in each room, the percentages stop changing.
    - You find that no matter what time of afternoon it is, roughly 45% are in the Library, 35% are in the Café, and 20% are in the Gym.
That stable balance is the limit distribution:$$\pi = [0.45, \, 0.35, \, 0.20]$$

**Intuition behind erodicity:** If you track just one single person walking through the building for 10,000 hours, they will spend roughly 45% of their total life in the Library, 35% in the Café, and 20% in the Gym. The limit distribution $\pi_i$ is simply the fraction of time a single trajectory spends in state $i$.

When you raise the transition matrix $P$ to a very large power $n$, every row converges to identical copies of $\pi$:$$P^n \xrightarrow{n \to \infty} \begin{bmatrix} \pi_1 & \pi_2 & \dots & \pi_K \\ \pi_1 & \pi_2 & \dots & \pi_K \\ \vdots & \vdots & \ddots & \vdots \\ \pi_1 & \pi_2 & \dots & \pi_K \end{bmatrix}$$
- Row 1: If you started in state 1, the probability of being in state $j$ after many steps is $\pi_j$.
- Row 2: If you started in state 2, the probability of being in state $j$ after many steps is $\pi_j$.

## Reversible HIA chain
An HIA Markov chain where the probabilistic dynamics look completely indistinguishable whether you observe the process running forwards in time or backwards in time.

An HIA Markov chain with transition matrix $P$ and stationary distribution $\pi$ is time-reversible if and only if it satisfies the detailed balance condition for every pair of states $(s_i, s_j) \in \mathcal{S}$:$$\pi_i P_{ij} = \pi_j P_{ji} \quad \text{for all } i, j$$
- **$\pi_i P_{ij}$ (Forward flux):** The unconditional probability of being at state $s_i$ at time $t$ and stepping into state $s_j$ at time $t+1$.
- **$\pi_j P_{ji}$ (Backward flux):** The unconditional probability of being at state $s_j$ at time $t$ and stepping into state $s_i$ at time $t+1$.

Reversibility means that in equilibrium, the probability flow along any individual path from $s_i$ to $s_j$ is balanced by an identical counter-flow from $s_j$ to $s_i$.

## Example

```python
import numpy as np
import matplotlib.pyplot as plt

# Define a transition matrix for a uniform distribution over three states
P = np.array([
    [0.0, 0.5, 0.5],
    [0.5, 0.0, 0.5],
    [0.5, 0.5, 0.0]
])

# Initial distribution (starting in state 0)
initial_distribution = np.array([1.0, 0.0, 0.0])
history = [initial_distribution]

# Simulate the Markov chain for a number of steps
n_steps = 20
for _ in range(n_steps):
    current_distribution = history[-1]
    next_distribution = np.dot(current_distribution, P)
    history.append(next_distribution)


# Theoretical stationary distribution (Uniform)
pi = np.array([1/3, 1/3, 1/3])

# Plot the evolution of the distribution
plt.figure(figsize=(10, 6))
for i in range(3):
    plt.plot([h[i] for h in history], label=f'State {i}')
plt.axhline(pi[0], color='gray', linestyle='--', label='Stationary Distribution')
plt.xlabel('Time Step')
plt.ylabel('Probability')
plt.title('Evolution of Markov Chain Distribution')
plt.legend()
plt.grid(True)
plt.show()
```
