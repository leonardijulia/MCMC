## Sampling a Random Variable

To sample a random variable (RV) means using a computer's uniform random number generator $u \sim \text{Uniform}(0, 1)$ to produce a concrete number or label that obeys a specific probability distribution. 

## The Main Sampling Methods

### Discrete Sampling Methods
- **Discrete Interval Carving (The Inversion Analogue for Discrete Variables)Principle**: Partition the unit interval $[0, 1]$ into contiguous segments with lengths matching the probability mass $p_k$ of each possible outcome.

### Continuous Sampling Methods
- **The Inversion Method (Inverse CDF)**: iven CDF $F_X(x) = P(X \le x)$, solve for $x = F_X^{-1}(u)$ with $u \sim \text{Uniform}(0, 1)$. Requires a closed-form, invertible CDF. The standard normal distribution $\mathcal{N}(0, 1)$ lacks a closed-form CDF.
- **Coordinate Transformation (Box-Muller Method)**: Avoids the lack of an analytical Gaussian 1D CDF by transitioning to 2D Cartesian space and switching to polar coordinates.
- **Grid Sampler**: Approximates an arbitrary continuous distribution in $(0, 1)$ by constructing its binary representation $y = \sum_{i=1}^N b_i \frac{1}{2^i}$ bit by bit. At each bit step $i$, calculate the conditional probability of landing in the lower half of the active interval and test against a uniform draw $u_i$ to assign $b_i \in \{0, 1\}$. 
- **Acceptance-Rejection Sampling**: Samples an unnormalized or complex target PDF $f_X(x)$ using an envelope proposal density $f_{X^*}(x)$ scaled by a constant $M \ge 1$ such that $f_X(x) \le M f_{X^*}(x)$ everywhere.
    1. Draw candidate $x \sim f_{X^*}(x)$.  
    2. Compute acceptance ratio $\rho(x) = \frac{f_X(x)}{M f_{X^*}(x)}$.  
    3. Draw $u \sim \text{Uniform}(0, 1)$.  
    4. Accept $x$ if $u \le \rho(x)$; otherwise reject and repeat.
