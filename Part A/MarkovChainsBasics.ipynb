## Markov Chains Introduction

### State Space & Stochastic Process
- **Stochastic Process**: A collection of random variables $\{X_t\}_{t \in T}$ indexed by time $t \in \{0, 1, 2, \dots\}$ representing the state of a system at each step.
- **State Space ($\mathcal{S}$)**: The set of all possible values (states) that $X_t$ can take. It can be discrete (finite or countably infinite, e.g., $\mathcal{S} = \{s_1, s_2, \dots, s_K\}$) or continuous ($\mathcal{S} \subseteq \mathbb{R}^d$).

### The Markov Property
A sequence of random variables $\{X_t\}$ is a Markov chain if the conditional probability of the future state depends only on the current state, completely independent of the past history: $$P(X_{t+1} = s_j \mid X_t = s_i, X_{t-1} = s_{t-1}, \dots, X_0 = s_0) = P(X_{t+1} = s_j \mid X_t = s_i)$$
- **Time-Homogeneity**: A Markov chain is homogeneous (stationary in time) if the transition probabilities do not change with time step $t$:$$P(X_{t+1} = s_j \mid X_t = s_i) = P_{ij} \quad \text{for all } t$$

A Markov Chain is a random process $\{X(t,w)\}$, where:
- t is discrete $t_n$, $n=0,1,... \rightarrow  X(t_n,w) = X_n(w)$ 
- for each n, X_n is a discrete and finite rv, whose possible values are called states i=1,...,m
- the marginal distribution of each $X_n$ is given by a vector $$\hat{p^{(n)}} = [p_1^{(n)}...p_m^{(n)}]^{T};  p_i^{(n)} \ge 0 \forall i,n; \sum_{i=1}^{m} p_i^{(n)} = 1$$

### Transition Probability Matrix & Kernel
- **Transition Matrix ($P$)**: For a discrete state space of size $K$, the transition probabilities are organized into a $K \times K$ stochastic matrix: $$P = \begin{bmatrix} P_{11} & P_{12} & \dots & P_{1K} \\ P_{21} & P_{22} & \dots & P_{2K} \\ \vdots & \vdots & \ddots & \vdots \\ P_{K1} & P_{K2} & \dots & P_{KK} \end{bmatrix}, \quad P_{ij} = P(X_{t+1} = s_j \mid X_t = s_i)$$
    - **Row-Stochastic Condition**: Every row sums to $1$ because moving somewhere in $\mathcal{S}$ is guaranteed: $$\sum_{j \in \mathcal{S}} P_{ij} = 1 \quad \text{and} \quad P_{ij} \ge 0$$
- **Transition Kernel ($T(x, dy)$ or $K(x, x')$)**: For continuous state spaces, the discrete matrix is replaced by a probability density kernel defining the transition density from point $x$ to $x'$: $$\int_\mathcal{S} K(x, x') \, dx' = 1$$
